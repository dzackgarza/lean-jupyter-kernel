# NbDsl — a Lean 4 elaborated DSL in Jupyter

Every cell below is ordinary **Lean 4**, elaborated by a persistent
`nbdsl_worker` process against the `NbDsl.Notebook` prelude. The DSL
commands (`let … ∈ …`, `prefer`, `#home`, `#via`) are Lean *command
elaborators* — the kernel never interprets them, and the declarations
they produce are checked by Lean's kernel like any other.

**All of mathlib is in scope**, `sage.all`-style.

```lean4
#eval 1 + 1
```

```lean4
open NbDsl NbDsl.Std
```

```lean4
let G := GrpCat.of PUnit ∈ Groups
```

```lean4
#check G
#home G
```

```lean4
prefer groupsToSets
```

```lean4
#via G ∈ Sets
```
