# NbDsl — a Lean 4 elaborated DSL in Jupyter

Every cell below is ordinary Lean, elaborated by a persistent `nbdsl_worker`
process against the `NbDsl.Notebook` prelude. The DSL commands (`let … ∈ …`,
`prefer`, `#home`, `#via`) are Lean command elaborators — the kernel never
interprets them, and the declarations they produce are checked by Lean's
kernel like any other.

```lean4
#eval 1 + 1
```

```lean4
open NbDsl NbDsl.Std
```

```lean4
prefer groupsToSets
```

```lean4
let G := GrpCat.of PUnit ∈ Groups
```

```lean4
#home G
```

```lean4
#via G ∈ Sets
```

```lean4
theorem two_and_two : 2 + 2 = 4 := by rfl
#check two_and_two
```
